```python
import pandas as pd
graph={
    "type_of_house":["flat","pg","flat","pg","flat"],
    "area_of_house":[55.25,77.89,82.33,49.44,50.00],
    "house_size":[2,5,10,1,2],
    "cost":[50000,80000,100000,45000,50000],
}
df=pd.DataFrame(graph)
df
```
